# Separate flow and reaction timescales

MRST retains microbial kinetics. Each coarse flow step is followed by local reaction substeps, with PHREEQC aqueous equilibrium after each one. The equation of state owns gas–liquid partitioning.

```matlab
startupH2sim
summary = exampleSequentialBiochemistryPhreeqc1D( ...
    'maximumFlowSteps', 2, 'plotResults', true, ...
    'phreeqcReflashParallel', false);
state = summary.states{end};
assert(all(state.sequentialH2BiochemPhreeqcReflashPass, 'all'));
plotSequentialBiochemistryPhreeqc(summary);
```

## Read the profiles



The default flow step is two days and the maximum local reaction substep is 0.4 days. PHREEQC inputs contain no `RATES` or `KINETICS` blocks. Serial pressure flashes do not require Parallel Computing Toolbox.

The current equilibrium-only backend has its elemental audit disabled. Its EOS inventory check does not by itself establish elemental conservation through the chemistry handoff.
